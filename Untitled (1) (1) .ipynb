```python
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import BertTokenizer, BertModel
from torch import nn
from sklearn.metrics import classification_report
from torch.optim import AdamW
from sklearn.model_selection import KFold
import numpy as np
from pypinyin import lazy_pinyin, Style
from cnradical import Radical, RunOption
import OpenHowNet

# Download OpenHowNet data
OpenHowNet.download()

# Data Preprocessing
df_train = pd.read_excel('usual_train.xlsx')
df_test = pd.read_excel('usual_test_labeled.xlsx')
df_train.columns = ['数据编号', 'text', 'label']
df_test.columns = ['数据编号', 'text', 'label']

label_mapping = {'happy': 0, 'sad': 1, 'neutral': 2, 'fear': 3, 'surprise': 4, 'angry': 5}
df_train['label'] = df_train['label'].map(label_mapping)
df_test['label'] = df_test['label'].map(label_mapping)

# Build pinyin and radical vocabularies
radical = Radical(RunOption.Radical)
all_texts = pd.concat([df_train['text'], df_test['text']]).unique()

# Pinyin vocabulary
pinyin_vocab = {}
pinyin_idx = 1
for text in all_texts:
    pinyins = lazy_pinyin(text, style=Style.TONE3)
    for pinyin in pinyins:
        if pinyin not in pinyin_vocab:
            pinyin_vocab[pinyin] = pinyin_idx
            pinyin_idx += 1

# Radical vocabulary
radical_vocab = {}
radical_idx = 1
for text in all_texts:
    for char in text:
        rad = radical.trans_ch(char) or ''
        if rad not in radical_vocab:
            radical_vocab[rad] = radical_idx
            radical_idx += 1

# Load HowNet data
hownet_dict = OpenHowNet.HowNetDict(init_sim=True)

def get_hownet_sememe_features(word, embed_dim=256):
    senses = hownet_dict.get_sense(word)
    if not senses:
        return np.zeros(embed_dim)

    sememes = set()
    for sense in senses:
        sememes.update(sense.get_sememe_list())

    sememe_vectors = [np.random.normal(size=embed_dim) for _ in sememes]
    return np.mean(sememe_vectors, axis=0) if sememe_vectors else np.zeros(embed_dim)

# Cache HowNet features
def build_hownet_cache(texts, embed_dim=256):
    cache = {}
    for text in texts:
        for char in text:
            if char not in cache:
                cache[char] = get_hownet_sememe_features(char, embed_dim)
    return cache

hownet_cache = build_hownet_cache(all_texts)

# Dataset class
class SentimentDataset(Dataset):
    def __init__(self, df, tokenizer, max_len=160):
        self.df = df
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.radical = Radical(RunOption.Radical)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]
        text = row['text']
        label = row['label']

        # Pinyin features
        pinyin = lazy_pinyin(text, style=Style.TONE3)
        pinyin_ids = [pinyin_vocab.get(p, 0) for p in pinyin]
        pinyin_ids = pinyin_ids[:self.max_len] + [0] * (self.max_len - len(pinyin_ids))

        # Radical features
        radicals = [self.radical.trans_ch(char) or '' for char in text]
        radical_ids = [radical_vocab.get(r, 0) for r in radicals]
        radical_ids = radical_ids[:self.max_len] + [0] * (self.max_len - len(radical_ids))

        # HowNet features
        hownet_features = [hownet_cache.get(char, np.zeros(256)) for char in text]
        hownet_features = np.array(hownet_features[:self.max_len])
        if hownet_features.shape[0] < self.max_len:
            padding = np.zeros((self.max_len - hownet_features.shape[0], 256))
            hownet_features = np.vstack([hownet_features, padding])

        hownet_features = torch.tensor(hownet_features, dtype=torch.float32)

        # Text encoding
        encoding = self.tokenizer.encode_plus(
            text,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt',
        )

        return {
            'text_input_ids': encoding['input_ids'].squeeze(0),
            'text_attention_mask': encoding['attention_mask'].squeeze(0),
            'pinyin_ids': torch.tensor(pinyin_ids, dtype=torch.long),
            'radical_ids': torch.tensor(radical_ids, dtype=torch.long),
            'hownet_embeddings': hownet_features,
            'labels': torch.tensor(label, dtype=torch.long),
        }

# MultiKernelCNN
class MultiKernelCNN(nn.Module):
    def __init__(self, embed_dim, out_channels, kernel_sizes):
        super(MultiKernelCNN, self).__init__()
        self.convs = nn.ModuleList([
            nn.Conv1d(in_channels=embed_dim, out_channels=out_channels, kernel_size=k) for k in kernel_sizes
        ])
        self.pool = nn.AdaptiveMaxPool1d(1)

    def forward(self, x):
        conv_results = [self.pool(torch.relu(conv(x))).squeeze(-1) for conv in self.convs]
        return torch.cat(conv_results, dim=1)

# SentimentModel Definition
class SentimentModel(nn.Module):
    def __init__(self, bert_model_path, dropout_prob=0.7, embed_dim=128, bigru_hidden_size=512, cnn_out_channels=64, kernel_sizes=(3, 4, 5)):
        super(SentimentModel, self).__init__()
        self.bert_text = BertModel.from_pretrained(bert_model_path)
        self.pinyin_embedding = nn.Embedding(len(pinyin_vocab) + 1, embed_dim)
        self.radical_embedding = nn.Embedding(len(radical_vocab) + 1, embed_dim)
        # BIGRU
        self.bigru = nn.GRU(768 + 256, bigru_hidden_size, num_layers=2, bidirectional=True, batch_first=True)
        self.bigru_ln = nn.LayerNorm(bigru_hidden_size * 2)
        self.bigru_dropout = nn.Dropout(dropout_prob)
        # MultiKernel CNN
        self.cnn_pinyin = MultiKernelCNN(embed_dim, cnn_out_channels, kernel_sizes)
        self.cnn_radical = MultiKernelCNN(embed_dim, cnn_out_channels, kernel_sizes)
        self.dropout = nn.Dropout(dropout_prob)
        # Recalculate input_dim
        input_dim = bigru_hidden_size * 2 + cnn_out_channels * len(kernel_sizes) * 2
        self.fc = nn.Linear(input_dim, 6)

    def forward(self, text_input_ids, text_attention_mask, pinyin_ids, radical_ids, hownet_embeddings):
        text_output = self.bert_text(input_ids=text_input_ids, attention_mask=text_attention_mask).last_hidden_state

        # Pinyin features
        pinyin_embedded = self.pinyin_embedding(pinyin_ids).permute(0, 2, 1)
        pinyin_features = self.cnn_pinyin(pinyin_embedded)

        # Radical features
        radical_embedded = self.radical_embedding(radical_ids).permute(0, 2, 1)
        radical_features = self.cnn_radical(radical_embedded)

        # BIGRU features with residual connection
        gru_input = torch.cat((text_output, hownet_embeddings), dim=2)
        gru_output, _ = self.bigru(gru_input)
        gru_output = gru_output + gru_input  # Add residual connection
        gru_output = self.bigru_ln(gru_output)
        gru_output = self.bigru_dropout(gru_output)
        gru_features = torch.cat((gru_output[:, -1, :gru_output.shape[2] // 2],
                                 gru_output[:, -1, gru_output.shape[2] // 2:]), dim=1)

        # Concatenate features
        combined_output = torch.cat((gru_features, pinyin_features, radical_features), dim=1)
        combined_output = self.dropout(combined_output)
        # Fully connected layer
        return self.fc(combined_output)

# Focal Loss
class FocalLoss(nn.Module):
    def __init__(self, gamma=2, alpha=None):
        super(FocalLoss, self).__init__()
        self.gamma = gamma
        self.alpha = alpha

    def forward(self, inputs, targets):
        ce_loss = nn.CrossEntropyLoss(reduction='none')(inputs, targets)
        probs = torch.softmax(inputs, dim=1)
        target_probs = probs.gather(1, targets.unsqueeze(1)).squeeze(1)
        loss = (1 - target_probs) ** self.gamma * ce_loss
        return loss.mean()

# Training and Evaluation Functions
def train(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0
    for batch in loader:
        optimizer.zero_grad()
        logits = model(
            batch['text_input_ids'].to(device),
            batch['text_attention_mask'].to(device),
            batch['pinyin_ids'].to(device),
            batch['radical_ids'].to(device),
            batch['hownet_embeddings'].to(device)
        )
        loss = criterion(logits, batch['labels'].to(device))
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)

def evaluate(model, loader, device):
    model.eval()
    predictions, true_labels = [], []
    with torch.no_grad():
        for batch in loader:
            logits = model(
                batch['text_input_ids'].to(device),
                batch['text_attention_mask'].to(device),
                batch['pinyin_ids'].to(device),
                batch['radical_ids'].to(device),
                batch['hownet_embeddings'].to(device)
            )
            predictions.extend(torch.argmax(logits, dim=1).cpu().numpy())
            true_labels.extend(batch['labels'].cpu().numpy())
    return predictions, true_labels

# Main program
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
tokenizer = BertTokenizer.from_pretrained('chinese-bert-wwm-ext')
criterion = FocalLoss()

kf = KFold(n_splits=10, shuffle=True, random_state=42)
best_model_path = './best_model.pth'
best_f1 = 0.0
avg_metrics = []

for fold, (train_idx, val_idx) in enumerate(kf.split(df_train)):
    train_dataset = SentimentDataset(df_train.iloc[train_idx], tokenizer)
    val_dataset = SentimentDataset(df_train.iloc[val_idx], tokenizer)
    train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=64)

    model = SentimentModel(bert_model_path='chinese-bert-wwm-ext').to(device)
    optimizer = AdamW(model.parameters(), lr=1e-5)

    for epoch in range(5):
        train_loss = train(model, train_loader, optimizer, criterion, device)
        print(f"Fold {fold + 1}, Epoch {epoch + 1}, Train Loss: {train_loss}")

    val_predictions, val_labels = evaluate(model, val_loader, device)
    metrics = classification_report(val_labels, val_predictions, target_names=label_mapping.keys(), output_dict=True, digits=4)
    avg_metrics.append(metrics['macro avg'])
    print(f"Fold {fold + 1} Classification Report:")
    print(classification_report(val_labels, val_predictions, target_names=label_mapping.keys(), digits=4))

    f1_score = metrics['macro avg']['f1-score']
    if f1_score > best_f1:
        best_f1 = f1_score
        torch.save(model.state_dict(), best_model_path)

# Print 10-fold average results
avg_f1 = np.mean([m['f1-score'] for m in avg_metrics])
print(f"10-Fold Average F1-Score: {avg_f1:.4f}")

# Load best model and test on test set
model.load_state_dict(torch.load(best_model_path))
test_dataset = SentimentDataset(df_test, tokenizer)
test_loader = DataLoader(test_dataset, batch_size=64)
test_predictions, test_labels = evaluate(model, test_loader, device)
print("Test Set Classification Report:")
print(classification_report(test_labels, test_predictions, target_names=label_mapping.keys(), digits=4))
```
